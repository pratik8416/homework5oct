Topic: List & Tuple 

1. Create a list named Total_student containing all student names.

2. Create a tuple named present_attendance containing today's present students.

3. Write a Python program to find and print the names of students who are absent today.

```python
Total_student = ["Raj", "Amit", "Jay", "Payal", "Priya","Rahul", "Pavan", "Kiran", "Prachi"]
present_attendance = ("Raj", "Jay", "Payal", "Priya", "Pavan", "Kiran")
absent_students = []
for student in Total_student:
    if student not in present_attendance:
        absent_students.append(student)
print("Absent Students")
print(absent_students)
```
